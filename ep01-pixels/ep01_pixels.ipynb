## 0. 準備：載入套件、下載像素腦的頭貼

```
import os, urllib.request
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

if not os.path.exists("mascot.png"):
    urllib.request.urlretrieve("https://raw.githubusercontent.com/pixelbraintw/pixelbrain-code/main/ep01-pixels/mascot.png", "mascot.png")

def show(*imgs, size=4):
    """把一張或多張圖並排顯示"""
    fig, axes = plt.subplots(1, len(imgs), figsize=(size * len(imgs), size))
    for ax, im in zip(np.atleast_1d(axes), imgs):
        ax.imshow(im, cmap="gray", vmin=0, vmax=255, interpolation="nearest")
        ax.axis("off")
    plt.show()
```

## 1. 把圖片讀成一堆數字

形狀是 (高, 寬, 通道)，每一格像素有 R、G、B 三個數字。

```
img = np.array(Image.open("mascot.png").convert("RGB"))
print("形狀 (高, 寬, 通道):", img.shape)
print("總共幾個數字:", img.size)
show(img)
```

## 2. 看看正中間那一格

```
h, w, _ = img.shape
print("正中間的像素 (R, G, B):", img[h // 2, w // 2])
```

## 3. 把眼睛放大來看

切出 64 × 64 的小區塊，平滑的眼睛會變回一格一格的方塊。

```
patch = img[191:255, 265:329]
show(patch, size=6)
print("反光那一格:", patch[19, 47])
```

## 4. 改數字就是改照片：左上角塗成紅色

```
red = img.copy()
red[:100, :100] = [255, 0, 0]
show(red)
```

## 5. 負片：用 255 減掉每個數字

```
negative = 255 - img
show(img, negative)
```

## 6. 轉成灰階

綠色權重最大，因為人眼對綠色最敏感。

```
R, G, B = img[..., 0], img[..., 1], img[..., 2]
gray = (0.299 * R + 0.587 * G + 0.114 * B).astype(np.uint8)
print("灰階後的形狀:", gray.shape)
show(img, gray)
```
